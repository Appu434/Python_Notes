```python
# List Data Type:
# List (list): An ordered collection of items, which can be of different data types.
# Lists are defined by square brackets [] and items are separated by commas.
# Lists are mutable, meaning their contents can be changed after creation.
# we can perform various operations on lists, such as adding, removing, and accessing elements.
# Example:
fruits = ["apple", "banana", "cherry"]
print(fruits)  # Output: ['apple', 'banana', 'cherry']
print(fruits[0])  # Accessing the first element: Output: apple
fruits.append("orange")  # Adding an element
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'orange']
fruits.remove("banana")  # Removing an element
print(fruits)  # Output: ['apple', 'cherry', 'orange']
# Lists can also contain other lists (nested lists).
nested_list = [1, 2, [3, 4], 5]
print(nested_list)  # Output: [1, 2, [3, 4], 5]
print(nested_list[2])  # Accessing the nested list: Output: [3, 4]
print(nested_list[2][0])  # Accessing an element in the nested list: Output: 3
# Lists support various methods for manipulation, such as sort(), reverse(), and pop(). 
numbers = [3, 1, 4, 1, 5]
numbers.sort()  # Sorting the list
print(numbers)  # Output: [1, 1, 3, 4, 5]
numbers.reverse()  # Reversing the list
print(numbers)  # Output: [5, 4, 3, 1, 1]
# Lists can be iterated over using loops.
for fruit in fruits:
    print(fruit)
# Output:
# apple
# cherry
# orange
# Lists can be sliced to create sublists.
sublist = fruits[1:3]
print(sublist)  # Output: ['cherry', 'orange']
# Lists can be concatenated using the + operator.
more_fruits = ["kiwi", "mango"]
all_fruits = fruits + more_fruits
print(all_fruits)  # Output: ['apple', 'cherry', 'orange', 'kiwi', 'mango']
```

```python
# Tuple Data Type:
# Tuple (tuple): An ordered, immutable collection of items, which can be of different data types.
# Tuples are defined by parentheses () and items are separated by commas.
# Tuples are immutable, meaning their contents cannot be changed after creation.
# we can perform various operations on tuples, such as accessing elements and unpacking.
# Example:
coordinates = (10.0, 20.0)
print(coordinates)  # Output: (10.0, 20.0)
print(coordinates[0])  # Accessing the first element: Output: 10.0
x, y = coordinates  # Unpacking the tuple
print(x, y)  # Output: 10.0 20.0
# Tuples can contain other tuples (nested tuples).
nested_tuple = (1, 2, (3, 4), 5)
print(nested_tuple)  # Output: (1, 2, (3, 4), 5)
print(nested_tuple[2])  # Accessing the nested tuple: Output: (3, 4)
print(nested_tuple[2][0])  # Accessing an element in the nested tuple: Output: 3
# Tuples support various methods for manipulation, such as count() and index().
numbers = (3, 1, 4, 1, 5)
print(numbers.count(1))  # Counting occurrences of 1: Output: 2
print(numbers.index(4))  # Finding the index of the first occurrence of 4: Output: 2
# Tuples can be iterated over using loops.
for number in numbers:
    print(number)
# Output:
# 3
# 1
# 4
# 1
# 5
# Tuples can be concatenated using the + operator.
more_numbers = (6, 7)
all_numbers = numbers + more_numbers
print(all_numbers)  # Output: (3, 1, 4, 1, 5, 6, 7)
```

```python
# Set Data Type:
# Set (set): An unordered collection of unique items, which can be of different data types
# Sets are defined by curly braces {} or the set() function, and items are separated by commas.
# Sets are mutable, meaning their contents can be changed after creation, but they do not allow duplicate items.
# we can perform various operations on sets, such as adding, removing, and performing set operations like union and intersection.
# Example:
unique_numbers = {1, 2, 3, 4, 5}
print(unique_numbers)  # Output: {1, 2, 3, 4, 5}
print(3 in unique_numbers)  # Checking membership: Output: True
unique_numbers.add(6)  # Adding an element
print(unique_numbers)  # Output: {1, 2, 3, 4, 5, 6}
unique_numbers.remove(2)  # Removing an element
print(unique_numbers)  # Output: {1, 3, 4, 5, 6}
# Sets do not allow duplicate items.
duplicate_set = {1, 2, 2, 3, 4, 4, 5}
print(duplicate_set)  # Output: {1, 2, 3, 4, 5}
# Sets support various set operations, such as union, intersection, difference, and symmetric difference.
set_a = {1, 2, 3}
set_b = {3, 4, 5}
union_set = set_a.union(set_b)  # Union
print(union_set)  # Output: {1, 2, 3, 4, 5}
intersection_set = set_a.intersection(set_b)  # Intersection
print(intersection_set)  # Output: {3}
difference_set = set_a.difference(set_b)  # Difference
print(difference_set)  # Output: {1, 2}
symmetric_difference_set = set_a.symmetric_difference(set_b)  # Symmetric Difference
print(symmetric_difference_set)  # Output: {1, 2, 4, 5}
# Sets can be iterated over using loops.
for number in unique_numbers:
    print(number)
# Output:
# 1
# 3
# 4
# 5
# Sets can be created from lists or tuples using the set() function.
list_with_duplicates = [1, 2, 2, 3, 4, 4, 5]
set_from_list = set(list_with_duplicates)   
print(set_from_list)  # Output: {1, 2, 3, 4, 5}
# Sets can be cleared using the clear() method.
unique_numbers.clear()
print(unique_numbers)  # Output: set()
```

```python
# Dictionary Data Type:
# Dictionary (dict): A collection of key-value pairs, where each key is unique and maps
# to a corresponding value. Dictionaries can store items of different data types.
# Dictionaries are defined by curly braces {} and key-value pairs are separated by commas.
# Dictionaries are mutable, meaning their contents can be changed after creation.
# we can perform various operations on dictionaries, such as adding, removing, and accessing elements.
# Example:
person = {"name": "Bob", "age": 30, "city": "New York"}
print(person)  # Output: {'name': 'Bob', 'age': 30, 'city': 'New York'}
print(person["name"])  # Accessing a value by key: Output: Bob
person["age"] = 31  # Updating a value
print(person)  # Output: {'name': 'Bob', 'age': 31, 'city': 'New York'}
person["job"] = "Engineer"  # Adding a new key-value pair
print(person)  # Output: {'name': 'Bob', 'age': 31, 'city': 'New York', 'job': 'Engineer'}
del person["city"]  # Removing a key-value pair
print(person)  # Output: {'name': 'Bob', 'age': 31, 'job': 'Engineer'}
# Dictionaries support various methods for manipulation, such as keys(), values(), items(), and get().
keys = person.keys()  # Getting all keys
print(keys)  # Output: dict_keys(['name', 'age', 'job'])
values = person.values()  # Getting all values
print(values)  # Output: dict_values(['Bob', 31, 'Engineer'])
items = person.items()  # Getting all key-value pairs
print(items)  # Output: dict_items([('name', 'Bob'), ('age', 31), ('job', 'Engineer')])
name = person.get("name")  # Getting a value using get()
print(name)  # Output: Bob
# Dictionaries can be iterated over using loops.
for key, value in person.items():
    print(f"{key}: {value}")
# Output:
# name: Bob
# age: 31
# Dictionaries can be nested, meaning a dictionary can contain other dictionaries as values.
employee = {
    "name": "Alice",
    "age": 28,
    "address": {
        "street": "123 Main St",
        "city": "Los Angeles",
        "state": "CA"
    }
}
print(employee)  # Output: {'name': 'Alice', 'age': 28, 'address': {'street': '123 Main St', 'city': 'Los Angeles', 'state': 'CA'}}
print(employee["address"]["city"])  # Accessing a nested value: Output: Los Angeles
# Dictionaries can be created using the dict() function.
person_dict = dict(name="Charlie", age=35, city="Chicago")
print(person_dict)  # Output: {'name': 'Charlie', 'age': 35, 'city': 'Chicago'}
# Dictionaries can be cleared using the clear() method.
person.clear()
print(person)  # Output: {}
```

```python
# Python Conditionals: Conditionals are used to perform different actions based on different conditions. The main conditional statements in Python are if, elif, and else.
# Example:
age = 20
if age < 18:
    print("You are a minor.")
elif age == 18:
    print("You are just an adult.")
else:
    print("You are an adult.")
    
```

```python
# Python Loops: Loops are used to execute a block of code multiple times. The main types of loops in Python are for and while loops.
# Example of a for loop:
for i in range(5):
    print(i)
# Example of a while loop:
count = 0
while count < 5:
    print(count)
    count += 1
```

```python
# If-else Statements: If-else statements are used to execute a block of code if a specified condition is true, and another block of code if the condition is false.
# Example:
number = 10
if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

```python
# elif Statements: The elif statement is used to check multiple conditions. It stands for "else if" and allows you to specify additional conditions if the previous if or elif conditions were false.
# Example:
number = 15
if number < 10:
    print("The number is less than 10.")
elif number == 10:
    print("The number is exactly 10.")
elif number < 20:
    print("The number is between 11 and 19.")
else:
    print("The number is 20 or greater.")
```

```python
# Nested If Statements: Nested if statements are if statements that are contained within another if statement. This allows for more complex decision-making processes.  
# Example:
age = 22
if age > 18:
    print("You are an adult.")
    if age > 21:
        print("You are over 21.")
    else:
        print("You are 18 or 21.")
else:
    print("You are a minor.")
```

```python
# Multiple Conditions and Logical Operators: You can combine multiple conditions using logical operators such as and, or, and not.
age = 22
if age > 18 and age < 30:
    print("You are a young adult.")
elif age >= 30 and age < 60:
    print("You are an adult.")
else:
    print("You are a senior.")
# Example with 'or' operator:
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday.")
# Example with 'not' operator:
is_raining = False
if not is_raining:
    print("It's a sunny day!")
# Example combining multiple logical operators:
age = 25
if age > 18 and not is_raining:
    print("You are a young adult enjoying a sunny day!")
    
```

```python
# For Loop: A for loop is used to iterate over a sequence (like a list, tuple, or string) or other iterable objects. It allows you to execute a block of code for each item in the sequence.
# Example:
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("I like", fruit)  
# Output:
# I like apple
# I like banana
# I like cherry
```

```python
# While Loop: A while loop is used to execute a block of code as long as a specified condition is true. It continues to loop until the condition becomes false.
# Example:
count = 0
while count < 5:
    print("Count is:", count)
    count += 1  
# Output:
```

```python
# Loop Control Statements: Loop control statements are used to modify the behavior of loops. The main loop control statements in Python are break, continue, and pass.
# Break Statement: The break statement is used to exit a loop prematurely when a certain condition is met.
# Example:
for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)
# Output:
# 0 
# 1
# 2
# 3
# 4
# Continue Statement: The continue statement is used to skip the current iteration of a loop and move to the next iteration.
# Example:
for i in range(10):
    if i % 2 == 0:
        continue  # Skip even numbers
    print(i)
# Output:
# 1
# 3
# 5
# 7
# 9 
# Pass Statement: The pass statement is a placeholder that does nothing. It is used when a statement is syntactically required but no action is needed.
# Example:
for i in range(5):
    if i == 3:
        pass  # Do nothing
    print(i)
# Output:
# 0
# 1
# 2
# 3
# 4
```

```python
# List append, extend, insert, remove, pop, clear, index, count, sort, reverse, copy
# Example:
my_list = [1, 2, 3]
my_list.append(4)  # Adds 4 to the end of the list
print(my_list)  # Output: [1, 2, 3, 4]
my_list.extend([5, 6])  # Adds multiple elements to the end of the list
print(my_list)  # Output: [1, 2, 3, 4, 5, 6]
my_list.insert(2, 2.5)  # Inserts 2.5 at index 2
print(my_list)  # Output: [1, 2, 2.5, 3, 4, 5, 6]
my_list.remove(3)  # Removes the first occurrence of 3
print(my_list)  # Output: [1, 2, 2.5, 4, 5, 6]
popped_element = my_list.pop()  # Removes and returns the last element
print(popped_element)  # Output: 6
print(my_list)  # Output: [1, 2, 2.5, 4, 5]
my_list.clear()  # Removes all elements from the list
print(my_list)  # Output: []
my_list = [3, 1, 4, 1, 5]
my_list.sort()  # Sorts the list in ascending order
print(my_list)  # Output: [1, 1, 3, 4, 5]
my_list.reverse()  # Reverses the order of the list
print(my_list)  # Output: [5, 4, 3, 1, 1]
my_list = [1, 2, 2, 3, 4]
count_of_twos = my_list.count(2)  # Counts occurrences of 2
print(count_of_twos)  # Output: 2
index_of_three = my_list.index(3)  # Finds the index of the first occurrence of 3
print(index_of_three)  # Output: 3
# Copying a list
original_list = [1, 2, 3]
copied_list = original_list.copy()  # Creates a shallow copy of the list
print(copied_list)  # Output: [1, 2, 3]
```

```python
# Frozen Set Data Type:
# Frozenset (frozenset): An unordered collection of unique items, similar to a set, but immutable (cannot be changed after creation).               
# Frozensets are defined using the frozenset() function and can contain items of different data types.
# we can perform various set operations on frozensets, such as union, intersection, difference, and symmetric difference.
# Example:
frozen_set_a = frozenset([1, 2, 3])
frozen_set_b = frozenset([3, 4, 5])
print(frozen_set_a)  # Output: frozenset({1, 2, 3})
print(frozen_set_b)  # Output: frozenset({3, 4, 5})
union_frozenset = frozen_set_a.union(frozen_set_b)  # Union
print(union_frozenset)  # Output: frozenset({1, 2, 3, 4, 5})
intersection_frozenset = frozen_set_a.intersection(frozen_set_b)  # Intersection
print(intersection_frozenset)  # Output: frozenset({3})
difference_frozenset = frozen_set_a.difference(frozen_set_b)  # Difference
print(difference_frozenset)  # Output: frozenset({1, 2})
symmetric_difference_frozenset = frozen_set_a.symmetric_difference(frozen_set_b)  # Symmetric Difference
print(symmetric_difference_frozenset)  # Output: frozenset({1, 2, 4, 5})
```

```python
# List Comprehensions: List comprehensions provide a concise way to create lists. They consist of brackets containing an expression followed by a for clause, and can include optional if clauses to filter items.
# Example:
squares = [x**2 for x in range(10)]  # Creates a list
print(squares)  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
even_squares = [x**2 for x in range(10) if x % 2 == 0]  # Creates a list of squares of even numbers
print(even_squares)  # Output: [0, 4, 16, 36, 64]
# Dictionary Comprehensions: Dictionary comprehensions provide a concise way to create dictionaries. They consist of curly braces containing a key-value expression followed by a for clause, and can include optional if clauses to filter items.
# Example:
squared_dict = {x: x**2 for x in range(10)}  # Creates a dictionary
print(squared_dict)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16, 5: 25, 6: 36, 7: 49, 8: 64, 9: 81}
# set Comprehensions: Set comprehensions provide a concise way to create sets. They consist of curly braces containing an expression followed by a for clause, and can include optional if clauses to filter items.
# Example:
squared_set = {x**2 for x in range(10)}  # Creates a set
print(squared_set)  # Output: {0, 1, 64, 4, 36, 9, 16, 49, 25, 81}
even_squared_set = {x**2 for x in range(10) if x % 2 == 0}  # Creates a set of squares of even numbers
print(even_squared_set)  # Output: {0, 64, 4, 36, 16}
# Tuple comprenensions: Tuple comprehensions are similar to list comprehensions but create tuples instead of lists. However, since tuples are immutable, tuple comprehensions are typically created using generator expressions.
# Example:
squared_tuple = tuple(x**2 for x in range(10))  # Creates a tuple
print(squared_tuple)  # Output: (0, 1, 4, 9, 16, 25, 36, 49, 64, 81)
```

```python
# Python Functions: Functions are reusable blocks of code that perform a specific task. They are defined using the def keyword, followed by the function name and parentheses.
# Some common functions: 
def greet(name):
    return f"Hello, {name}!"

def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
    
def is_even(num):
    return num % 2 == 0

def is_odd(num):
    return num % 2 != 0

def fibonacci(n):
    a, b = 0, 1
    sequence = []
    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b
    return sequence

def power(base, exponent):
    return base ** exponent

def square(num):
    return num * num

def cube(num):
    return num * num * num

def average(numbers):
    return sum(numbers) / len(numbers)

def max_in_list(numbers):
    return max(numbers)

def min_in_list(numbers):
    return min(numbers)

```

```python
# How to call a function ?:
print(greet("Alice"))  # Output: Hello, Alice!
print(add(5, 3))  # Output: 8
print(multiply(4, 2))  # Output: 8
print(factorial(5))  # Output: 120
print(power(2, 3))  # Output: 8
print(square(4))  # Output: 16
print(cube(3))  # Output: 27
print(average([1, 2, 3, 4, 5]))  # Output: 3.0
print(max_in_list([1, 2, 3, 4, 5]))  # Output: 5
print(min_in_list([1, 2, 3, 4, 5]))  # Output: 1
```

```python
# Function Arguments: Functions can take arguments (inputs) to perform operations. There are different types of arguments:
# Positional Arguments: Arguments that are passed to a function in the correct positional order.
# Keyword Arguments: Arguments that are passed to a function by explicitly specifying the parameter name.
# Default Arguments: Arguments that take a default value if no value is provided during the function call.
# Code Example:
# Positional Arguments
def describe_person(name, age=30, city="New York"):
    return f"{name} is {age} years old and lives in {city}."
# Keyword Arguments
print(describe_person(age=25, name="Alice", city="Los Angeles"))  # Output: Alice is 25 years old and lives in Los Angeles.
# Default Arguments
print(describe_person("Bob"))  # Output: Bob is 30 years old and lives in New York.
```

```python
# LAMBDA Functions: Lambda functions are small anonymous functions defined using the lambda keyword. They can take any number of arguments but can only have a single expression. The expression is evaluated and returned when the function is called.
# Syntax: lambda arguments: expression
# Example:
add = lambda x, y: x + y
print(add(5, 3))  # Output: 8
multiply = lambda x, y: x * y
print(multiply(4, 2))  # Output: 8
factorial = lambda n: 1 if n == 0 else n * factorial(n-1)
print(factorial(5))  # Output: 120
# Lambda sorting example:
points = [(1, 2), (3, 1), (5, 4), (2, 3)]
points.sort(key=lambda point: point[1])  # Sort by the second element of each tuple
print(points)  # Output: [(3, 1), (1, 2), (2, 3), (5, 4)]
# Lambda with filter example:
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print(even_numbers)  # Output: [2, 4, 6]
# Lambda with map example:
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25, 36]
```

```python
# Scope and Lifetime of Variables: In Python, the scope of a variable refers to the region of the code where the variable is accessible. The lifetime of a variable refers to the duration for which the variable exists in memory. Variables can have different scopes based on where they are defined:
# Local Scope: Variables defined inside a function are in the local scope and can only be accessed within that function.
# Global Scope: Variables defined outside of any function are in the global scope and can be accessed from anywhere in the code.
# Nonlocal Scope: Variables defined in an enclosing function can be accessed in nested functions using the nonlocal keyword.
# Example:
x = "global x"  # Global variable
def outer_function():
    y = "outer y"  # Enclosing variable
    def inner_function():
        z = "inner z"  # Local variable
        print(z)  # Accessing local variable
        print(y)  # Accessing enclosing variable
        print(x)  # Accessing global variable
    inner_function()
    print(y)  # Accessing enclosing variable
outer_function()
print(x)  # Accessing global variable
```

```python
# Local Variable Example:
def my_function():
    local_var = "I am local"
    print(local_var)  # Output: I am local
my_function()
# print(local_var)  # This will raise an error because local_var is not accessible outside the function
```

```python
# Global Variable Example:
global_var = "I am global"
def my_function():
    print(global_var)  # Accessing global variable
my_function()
print(global_var)  # Output: I am global
```

```python
# Nonlocal Variable Example:
def outer_function():
    nonlocal_var = "I am nonlocal"
    def inner_function():
        nonlocal nonlocal_var
        nonlocal_var = "I am modified nonlocal"
        print(nonlocal_var)  # Output: I am modified nonlocal
    inner_function()
    print(nonlocal_var)  # Output: I am modified nonlocal
```

```python
# Modify global variable Example:
counter = 0  # Global variable
def increment():
    global counter  # Declare counter as global to modify it
    counter += 1
increment()
print(counter)  # Output: 1
# Modify nonlocal variable Example:
```

```python
def outer():
    count = 0  # Enclosing variable
    def inner():
        nonlocal count  # Declare count as nonlocal to modify it
        count += 1
        print(count)  # Output: 1
    inner()
    print(count)  # Output: 1
outer()
```
