### Object-Oriented Programming (OOP) Concepts


#### 1. Classes and Objects
- **Class**: A blueprint for creating objects (a particular data structure), defining initial values (attributes) and behaviors (methods).
- **Object**: An instance of a class.

#### 2. Encapsulation
- Bundling the data (attributes) and methods (functions) that operate on the data into a single unit (class).
- Restricting access to certain components (private vs public).

#### 3. Inheritance
- Mechanism by which one class can inherit attributes and methods from another class.
- Promotes code reusability.

#### 4. Polymorphism
- Ability to present the same interface for different data types.
- Method overriding and method overloading are examples.

#### 5. Abstraction
- Hiding complex implementation details and showing only the essential features of the object.
- Achieved through abstract classes and interfaces.

```python
# Class and Object Example in Python
# Defining a class named 'watch'
class watch: 
    # Constructor to initialize attributes
    def __init__(self, brand, price, color): # __init__ is a special method called a constructor
        self.brand = brand # Attribute to store the brand of the watch
        self.price = price # Attribute to store the price of the watch
        self.color = color # Attribute to store the color of the watch

    def show_time(self): # Method to display the time
        print("The time is 10:10 AM") # Example method to show time

# Creating an object of the watch class
my_watch = watch("Rolex", 5000, "Black") # Instantiating an object with brand, price, and color

# Accessing attributes and methods
print(my_watch.brand)  # Output: Rolex
my_watch.show_time()   # Output: The time is 10:10 AM
```

```python
# what is class? in easy and understandable way
# A class is like a blueprint for creating objects. It defines the properties (attributes) and behaviors (methods) that the objects created from the class will have.
# For example, if you have a class called "Car", it might have attributes like "color", "make", and "model", and methods like "start_engine" and "stop_engine".
# When you create an object (or instance) of the class, you are creating a specific car with its own unique values for those attributes.
# In summary, a class is a template for creating objects, and an object is an instance of a class with specific values.
```

```python
# what is object? in easy and understandable way
# An object is a specific instance of a class. It is created based on the blueprint provided by the class.
# For example, if you have a class called "Dog", you can create multiple objects (instances) of that class, each representing a different dog with its own unique characteristics.
# For instance, you might create an object named "my_dog" with attributes like "name" set to "Buddy", "breed" set to "Golden Retriever", and "age" set to 3.
# Each object can also have its own behaviors (methods) defined by the class, such as "bark" or "fetch".
# In summary, an object is a specific example of a class, with its own unique values for the attributes defined in the class.
```

```python
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species

    def make_sound(self):
        if self.species == "Dog":
            print(f"{self.name} says Woof!")
        elif self.species == "Cat":
            print(f"{self.name} says Meow!")
        else:
            print(f"{self.name} makes a sound!")
# Creating objects of the Animal class
dog = Animal("Buddy", "Dog")
cat = Animal("Whiskers", "Cat")
dog.make_sound()  # Output: Buddy says Woof!
cat.make_sound()  # Output: Whiskers says Meow!
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print(f"The engine of the {self.year} {self.make} {self.model} is starting.")
    def stop_engine(self):
        print(f"The engine of the {self.year} {self.make} {self.model} is stopping.")
# Creating an object of the Car class
my_car = Car("Toyota", "Camry", 2020)
my_car.start_engine()  # Output: The engine of the 2020 Toyota Camry is starting.
my_car.stop_engine()   # Output: The engine of the 2020 Toyota Camry is stopping.
```

```python
# Inheritance:
# Inheritance is a fundamental concept in object-oriented programming (OOP) that allows a class (called the child or subclass) to inherit attributes and methods from another class (called the parent or superclass).
# This promotes code reusability and establishes a hierarchical relationship between classes.
# The child class can extend or override the functionality of the parent class, allowing for more specialized behavior.
# Inheritance is often described as an "is-a" relationship. For example, if you have a parent class called "Animal" and a child class called "Dog", you can say that a Dog "is an" Animal.
# This means that the Dog class inherits the properties and behaviors of the Animal class, while also adding its own unique features.
# In summary, inheritance allows classes to share and reuse code, making it easier to create and maintain complex systems.
# Example of Inheritance in Python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def start(self):
        print(f"The {self.make} {self.model} is starting.")

    def stop(self):
        print(f"The {self.make} {self.model} is stopping.")

# Creating a child class that inherits from Vehicle
class Car(Vehicle):
    def __init__(self, make, model, year):
        super().__init__(make, model) # super() is used to call the constructor of the parent class
        self.year = year

    def display_info(self):
        print(f"Car Information: {self.year} {self.make} {self.model}")

# Creating an object of the Car class
my_car = Car("Toyota", "Camry", 2020)
my_car.start()        # Output: The Toyota Camry is starting.
my_car.display_info() # Output: Car Information: 2020 Toyota Camry
my_car.stop()         # Output: The Toyota Camry is stopping.
```

```python
# Encapsulation:
# Encapsulation is a fundamental concept in object-oriented programming (OOP) that involves bundling the data (attributes) and methods (functions) that operate on that data into a single unit, typically a class.
# It restricts direct access to some of the object's components, which is a means of preventing unintended interference and misuse of the methods and data.
# This is often achieved by making some attributes or methods private (not accessible from outside the class) while providing public methods to access or modify them.
# Encapsulation helps to maintain the integrity of the data by controlling how it is accessed and modified.
# Example of Encapsulation in Python
class BankAccount:
    def __init__(self, account_number, initial_balance=0):
        self.account_number = account_number
        self.__balance = initial_balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: {amount}")
        else:
            print("Invalid withdrawal amount.")

    def get_balance(self):
        return self.__balance

# Creating an object of the BankAccount class
account = BankAccount("123456789", 1000)
account.deposit(500)          # Output: Deposited: 500
account.withdraw(200)         # Output: Withdrew: 200
print(account.get_balance())  # Output: 1300
```

```python
# Polymorphism:
# Polymorphism is a core concept in object-oriented programming (OOP) that allows objects of different classes to be treated as objects of a common superclass.
# It enables a single interface to represent different underlying forms (data types).
# Polymorphism allows methods to do different things based on the object it is acting upon, even if they share the same name.
# This is often achieved through method overriding (where a subclass provides a specific implementation of a method already defined in its superclass) and method overloading (where multiple methods have the same name but different parameters).
# Example of Polymorphism in Python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Creating a list of animals
animals = [Dog(), Cat()]

# Demonstrating polymorphism
for animal in animals:
    animal.speak()
    
```

```python
# Abstraction:
# Abstraction is a fundamental concept in object-oriented programming (OOP) that focuses on hiding complex implementation details and exposing only the essential features of an object.
# It allows programmers to reduce complexity and increase efficiency by providing a simplified interface to interact with objects.
# Abstraction is often achieved through abstract classes and interfaces, which define methods that must be implemented by derived classes, without specifying the exact implementation.
# This helps in separating the "what" from the "how", allowing developers to work at a higher level of understanding without needing to know the intricate details of the implementation.
# Example of Abstraction in Python
from abc import ABC, abstractmethod # From the abc module, we import ABC and abstractmethod to create abstract classes and methods

class Shape(ABC):
    @abstractmethod # Abstract method that must be implemented by subclasses
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

# Creating objects of the concrete classes
rect = Rectangle(10, 5)
circ = Circle(7)

print("Rectangle Area:", rect.area())
print("Circle Area:", circ.area())
```

```python
class CoffeeMachine: # Abstraction example 
    def start(self): # Public method to start the machine
        # inside here could be hundreds of steps
        print("Machine starting...")

    def brew_coffee(self): # Public method to brew coffee       
        # inside here could be hundreds of steps
        print("Brewing coffee...")

# User side
machine = CoffeeMachine()
machine.start()
machine.brew_coffee()
```

```python
from abc import ABC, abstractmethod # Abstract Base Class example

class Animal(ABC):        # abstract base class
    @abstractmethod
    def make_sound(self): # only says "there should be a method called make_sound"
        pass

class Dog(Animal):         # child implements it
    def make_sound(self):
        print("Bark!")

class Cat(Animal):         # child implements it
    def make_sound(self):
        print("Meow!")

# We cannot do Animal() directly, it's abstract
d = Dog()
c = Cat()

d.make_sound()  # Bark!
c.make_sound()  # Meow!
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    class Rectangle(Shape):
        def __init__(self, width, height):
            self.width = width
            self.height = height
            
        def area(self):
            return self.width * self.height

    class Circle(Shape):
        def __init__(self, radius):
            self.radius = radius
            
        def area(self):
            return 3.14 * self.radius * self.radius
        
# Object creation
rect = Rectangle(10, 5)
circ = Circle(7)
print("Rectangle Area:", rect.area())
print("Circle Area:", circ.area())


    
```

```python
# Decorators:
# Decorators are a powerful and expressive feature in Python that allows you to modify or enhance the behavior of functions or methods without changing their actual code.
# A decorator is essentially a function that takes another function as an argument, adds some functionality to it, and returns a new function.
# Decorators are often used for cross-cutting concerns such as logging, access control, caching, and performance measurement.
# They help in keeping the code clean and maintainable by separating concerns.  
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
def decorated_function(func):
    def wrapper(text):
        result = func(text)
        return result.upper()
    return wrapper

@decorated_function
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))
```
