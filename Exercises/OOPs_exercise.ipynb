Exercise 3: Build a class Employee with multiple constructors that can initialize an employee object in different ways.

```python
class Employee: # Class to represent an Employee
    def __init__(self,name,position,salary): # Constructor to initialize employee details 
        self.name = name
        self.position = position
        self.salary = salary
        
    def display_info(self): # Method to display employee information
        return f"Name: {self.name}, Position: {self.position}, Salary: ${self.salary}"
    
emp1 = Employee("Alice","Developer",70000) # Creating an instance of Employee
emp2 = Employee("Bob","Designer",65000)

print(emp1.display_info()) # Output: Name: Alice, Position: Developer, Salary: $70000
print(emp2.display_info()) # Output: Name: Bob, Position: Designer, Salary: $65000
```

Exercise 4: Design a class SeriesCalculator that calculates the sum of an arithmetic series.

```python
# Series Calculator that calculates the sum of an arithmetic series
# explaination: n = number of terms, a = first term, d = common difference 
# Example: For n=5, a=1, d=2 => Series: 1,3,5,7,9 => Sum = 25
# Formula derivation: Sum of n terms = n/2 * (2a + (n-1)d)
class SeriesCalculator:
    def sum(self, n, a=1, d=2):
        return n * (2 * a + (n - 1) * d) // 2
sc = SeriesCalculator()
print("Sum of series:", sc.sum(5)) # Output: 25
```

Exercise 5: Create a class MaxFinder that identifies the largest number in a list.

```python
# formula for finding largest number in the list is max()
# Example: numbers = [1, 2, 3, 4, 5] => Largest = 5
class MaxFinder:
    def find_max(self, numbers):
        return max(numbers)
    
mf = MaxFinder()
print("Largest number:", mf.find_max([1, 2, 3, 4, 5])) # Output: 5

print(mf.find_max([-10, -20, -3, -4, -5])) # Output: -3

print(mf.find_max([40,33,80,99,27,68,91,77,100,22])) # Output: 100
```

Exercise 4: Class Inheritance
Given:
Create a Bus class that inherits from the Vehicle class. Give the capacity argument of Bus.seating_capacity() a default value of 50.

```python
class Vehicle:
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

    def seating_capacity(self, capacity):
        return f"The seating capacity of a {self.name} is {capacity} passengers"
    
class Bus(Vehicle):
    def seating_capacity(self, capacity=50):
        return super().seating_capacity(capacity)
    
SchoolBus = Bus("School Volvo", 180, 12)
print(SchoolBus.seating_capacity()) # Output: The seating capacity of a School Volvo is 50 passengers
print(SchoolBus.seating_capacity(60)) # Output: The seating capacity of a School
```

Exercise 10: Calculate the area of different shapes using OOP
Given:

You have given a Shape class and subclasses Circle  and Square. The parent class (Shape) has a area() method.

Now, Write a OOP code to calculate the area of each shapes (each subclass must write its own implementation of area() method to calculates its area).

```python
class Shape:
    def area(self):
        raise NotImplementedError("Area method must be implemented by subclasses")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14159 * self.radius * self.radius

class Square(Shape):
    def __init__(self, side):
        self.side = side
        
    def area(self):
        return self.side * self.side

# Example of polymorphism
shapes = [Circle(5), Square(7), Circle(3)]

for shape in shapes:
    print(shape.area())  # Output: 78.53975, 49, 28.27431
```
